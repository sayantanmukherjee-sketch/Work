```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': [],
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return traversal

result = bfs(graph, 'A')

print("BFS Traversal:")
print("->".join(result))

G = nx.DiGraph(graph)

# for node in graph:
#     for neighbor in graph[node]:
#         G.add_rdgr(node, neighbor)

pos = {
     'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (-0.5, 0)
}
plt.figure(figsize=(8,6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2000,
    node_color="lightblue",
    font_size=12,
    font_weight="bold",
    arrows=True
)

plt.title("Graph Representation for BFS")
plt.show()
```
