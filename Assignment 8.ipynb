```python
import math

def minimax(depth,node_index,is_maximizing,values,max_depth):
    if depth==max_depth:
        return values[node_index]

    if is_maximizing:
        return max(
            minimax(depth+1,node_index*2,False,values,max_depth),
            minimax(depth+1,node_index*2+1,False,values,max_depth)
            
        )
    else:
        return min(
            minimax(depth+1,node_index*2,True,values,max_depth),
            minimax(depth+1,node_index*2+1,True,values,max_depth)
        )


leaf_values=[3,5,2,9,12,5,23,33]

Tree_depth=int(math.log2(len(leaf_values)))

best_score=minimax(0,0,True,leaf_values,Tree_depth)

print("Best possible score:",best_score)

```

```python
import math

# Alpha-Beta Pruning Algorithm
def alpha_beta(depth, node_index, maximizing_player, values, alpha, beta, max_depth):

    # If leaf node is reached
    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        best = float('-inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                alpha,
                beta,
                max_depth
            )

            best = max(best, value)
            alpha = max(alpha, best)

            # Beta cut-off
            if beta <= alpha:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                alpha,
                beta,
                max_depth
            )

            best = min(best, value)
            beta = min(beta, best)

            # Alpha cut-off
            if beta <= alpha:
                break

        return best


# Leaf node values
values = [3, 5, 2, 9, 12, 5, 23, 23]

max_depth = 3

alpha = float('-inf')
beta = float('inf')

# MAX starts the game
optimal_score = alpha_beta(
    0, 0, True, values, alpha, beta, max_depth
)

print("Leaf Node Values:", values)
print("Optimal Score for MAX player:", optimal_score)
           
```
